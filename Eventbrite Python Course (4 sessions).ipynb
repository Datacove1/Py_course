Variables can store data of different types, and different types can do different things.

Python has the following data types built-in by default, in these categories:

Text Type:
- str (String): Used to store text data. Strings are sequences of characters enclosed in single, double, or triple quotes.

Numeric Types:
- int (Integer): Used for whole numbers, e.g., 42.
- float (Floating-Point): Used for decimal or floating-point numbers, e.g., 3.14.
- complex: Used for complex numbers, e.g., 2 + 3j.

Sequence Types:
- list: Ordered, mutable collections that can store elements of different data types, e.g., [1, 2, 'three'].
- tuple: Ordered, immutable collections, e.g., (1, 2, 3).
- range: Represents an immutable sequence of numbers, often used for iteration, e.g., range(5).

Mapping Type:
- dict (Dictionary): An unordered collection of key-value pairs, where keys are unique, e.g., {'name': 'John', 'age': 30}.

Set Types:
- set: An unordered collection of unique elements, e.g., {1, 2, 3}.
- frozenset: Similar to set, but immutable.

Boolean Type:
- bool (Boolean): Represents either True or False, often used for conditional statements and comparisons.

Binary Types:
- bytes: Used to store sequences of bytes (immutable), e.g., b'hello'.
- bytearray: Similar to bytes but mutable.
- memoryview: Represents a memory view of an object's buffer.

None Type:
- NoneType: Represents the absence of a value or a null value, typically used to initialize variables or indicate missing data.

### Lists

Lists are used to store multiple items in a single variable by using [ ]. 

List items are 
- Ordered: this means that the items have a defined order, and that order will not change. If you add new items to a list, the new items will be placed at the end of the list.
- Changeable: meaning that we can change, add, and remove items in a list after it has been created.
- Allows duplicate values: since lists are indexed, lists can have items with the same value

Colours = red, blue, green, yellow, blue, pink

```python
# Create a list of favorite colours
colours = ["red", "blue", "green", "yellow", "blue", "pink"]

# Access your favorite colour using indexing
your_favorite_colour = colours[3]
print("Favorite colour:", your_favorite_colour)

# Reverse the order of the list using indexing
reversed_colours = colours[::-1]
print("Reversed list:", reversed_colours)

# Append "green" to the list
colours.append("green")
print("Append 'green':", colours)

# Insert "purple" in the second position
colours.insert(1, "purple")
print("Insert 'purple':", colours)

# Remove the third colour
colours.remove("yellow")
print("Removed yellow:", colours)

# Use pop() on the list
colours.pop()
print("List after pop:", colours)

# Sort the list in reverse alphabetical order
colours.sort(reverse=True)
print("Reverse alphabetical order:", colours)
```

### Python Conditions and If statements

```python
# Make an 'if' statement (we saw this in the last session)

a = 33
b = 200

if b > a:
    print("b is greater than a")
```

- When using if statements in Python we use the ':' to signify the start of a new code block
- The indentation (space before print) shows what code is inside the code block - very important to Python. Other langugaes may use {}
- If the condition is true then the statement will print

The colon : indicates that the following indented block of code is associated with the if statement. Without the colon and proper indentation, Python would not know where the if block starts and ends, and it would result in a syntax error. Therefore, both the colon and indentation are essential for structuring and defining the control flow in Python programs.

```python
# Another example

x = 7 + (3 * 5) - (4 ** 2)
y = ((4 * 17) + 4) / 12


if x == y:
    print("Both equations are equal")
    
# But what if it not True? 
# Nothing happens! So how can we solve this? We could just write another if statement;

# x = 10 + (3 * 5) - (4 ** 2)
# y = ((4 * 17) + 4) / 12

# if x != y: 
#     print("The equations are not equal")

# But there os a better way
```

```python
# To give two conditions we use 'if' and 'else' statements - all need indentation (use tab)

x = 10 + (3 * 5) - (4 ** 2)
y = ((4 * 17) + 4) / 12


if x == y:
    print("Both equations are equal")
else:
    print("The equations are not equal")
```

```python
# To give multiple conditions we use 'if', 'elif', and 'else' statements - all need indentation (use tab)

a = 330
b = 200

if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
else:
    print("a is greater than b")
```

#### Loops (for and while)

A 'for' loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).

This is less like the for keyword in other programming languages, and works more like an iterator method as found in other object-orientated programming languages.

With the for loop we can execute a set of statements, once for each item in a list, tuple, set etc.

```python
# We can loop through the characters in a string, with a 'for' loop

for letters in "banana":
    print(letters)
```

```python
# We can loop through a list

fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
```

```python
# We can add in a 'break' into the loop to stop it before it has looped through all the items

for x in fruits:
    print(x)
    if x == "banana":
    break
```

```python
# What do you think is going to happen differently here? 

fruits = ["apple", "banana", "cherry"]

for x in fruits:
    if x == "banana":
        break
    print(x)
```

The order of the instructions are important!

```python
# With the continue statement we can stop the current iteration of the loop, and continue with the next:

fruits = ["apple", "banana", "cherry"]

for x in fruits:
    if x == "banana":
        continue
    print(x)
```

To loop through a set of code a specified number of times, we can use the range() function.

The range() function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and ends at a specified number.

```python
for x in range(6):
    print(x)
```

#### Game - Guess the number!

Instructions:
- Generate a random number between 1 and 20.
- Ask the player to guess the number.
- If the player's guess is correct, print a congratulatory message and end the game.
- If the guess is incorrect, provide feedback (higher or lower) and allow the player to guess again.
- Repeat steps 3 and 4 until the player guesses the correct number.

```python
import random        

# Generate a random number between 1 and 20
secret_number = random.randint(1, 20)

print("Welcome to the Number Guessing Game!")
attempts = 0

# Use a while loop for the game
while True:
    guess = int(input("Guess the number (between 1 and 20): "))
    attempts += 1

    if guess == secret_number:
        print(f"Congratulations! You guessed the number {secret_number} correctly in {attempts} attempts.")
        break
    elif guess < secret_number:
        print("Try a higher number.")
    else:
        print("Try a lower number.")
```

In this example:

- We generate a random secret number using random.randint() between 1 and 20.
- We use a while loop with while True: to repeatedly ask the player for their guess until they guess the correct number.
- We keep track of the number of attempts with the attempts variable.
- If the player's guess is correct, we print a congratulatory message and break out of the loop.
- If the guess is incorrect, we provide feedback to help the player guess correctly and continue the loop.

This example demonstrates the use of a while loop to automate the repetitive task of guessing the number until the correct answer is found, which is a common use case for loops in programming.
